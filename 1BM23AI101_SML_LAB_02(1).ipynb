```python
import numpy as np
shear= [1.2, 2.5, 3.1, 4.0, 5.2, 6.3, 7.4, 8.1, 9.0, 10.2, 11.1, 12.4, 13.3, 14.5, 15.0, 16.2, 17.5, 18.1, 19.3, 20.0],
age=[21, 22, 24, 25, 28, 30, 31, 34, 35, 37, 40, 42, 43, 45, 48, 50, 51, 53, 55, 57]

def gradient_descent(x, y, initial_learning_rate=0.01, decay_rate=0.01, n_iterations=1000):
    x=np.array(x)
    y=np.array(y)
    X = np.c_[np.ones(len(age)), age]
    y = np.array(shear)
    m = len(y)  
    theta = np.random.randn(2)
    
    for iteration in range(n_iterations):
        predictions = x.dot(theta)
        gradients = (2/m) * x.T.dot(predictions - y)
        learning_rate = initial_learning_rate / (1 + decay_rate * iteration)
        theta -= learning_rate * gradients
        error = np.mean((predictions - y) ** 2)
        
        if error > 0.1:
            iteration +=0.1
        break
    return gradients
    return theta



theta_gd = gradient_descent(x, y)

print("In Gradient Descent:")
print(f"Intercept: {theta_gd[0]:.4f}, Slope: {theta_gd[1]:.4f}")
```
